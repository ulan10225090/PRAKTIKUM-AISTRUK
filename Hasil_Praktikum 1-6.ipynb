```python
# 1. Buat variabel harga dan jumlah
harga = 12500
jumlah = 4

# 2. Hitung total harga
total_harga = harga * jumlah

# 3. Hitung sisa pembagian total harga dengan 2.000 menggunakan operator %
sisa_bagi = total_harga % 2000

# 4. Hitung harga setelah diskon Rp5.000
harga_diskon = total_harga - 5000

# 5. Tampilkan hasil setiap perhitungan dengan label yang jelas
print("--- HASIL PERHITUNGAN TOKO BUKU ---")
print(f"Harga per buah        : Rp {harga}")
print(f"Jumlah beli           : {jumlah} buah")
print(f"Total Harga           : Rp {total_harga}")
print(f"Sisa Bagi (mod 2000)  : Rp {sisa_bagi}")
print(f"Harga Setelah Diskon  : Rp {harga_diskon}")
```

```python
# 1. Buat variabel nilai dan kehadiran
nilai = 82
kehadiran = 90

# Syarat kelulusan: nilai minimal 75 dan kehadiran minimal 80%
syarat_nilai = 75
syarat_kehadiran = 80

print(f"Data Awal -> Nilai: {nilai}, Kehadiran: {kehadiran}%")

# 2. Gunakan operator >= untuk memeriksa masing-masing syarat
lulus_nilai = nilai >= syarat_nilai
lulus_kehadiran = kehadiran >= syarat_kehadiran
print("Apakah memenuhi syarat nilai?     :", lulus_nilai)
print("Apakah memenuhi syarat kehadiran? :", lulus_kehadiran)

# 3. Gunakan operator and untuk menentukan kelulusan
lulus_semua_and = lulus_nilai and lulus_kehadiran
print("Status Kelulusan (AND)           :", lulus_semua_and)

# 4. Ubah nilai menjadi 70
print("\n--- SETELAH NILAI DIUBAH MENJADI 70 ---")
nilai = 70
lulus_nilai = nilai >= syarat_nilai
print("Nilai baru:", nilai)
print("Status Kelulusan Baru (AND)      :", lulus_nilai and lulus_kehadiran)

# 5. Coba gunakan or
lulus_semua_or = lulus_nilai or lulus_kehadiran
print("Status Kelulusan dengan OR       :", lulus_semua_or)
```

```python
def rekap_nilai(nama_mhs, tugas, ujian):
    # 3. Hitung rata-rata kedua nilai
    rata_rata = (tugas + ujian) / 2

    # 4. Status lulus jika rata-rata minimal 75
    status_lulus = rata_rata >= 75

    # 5. Tampilkan hasil
    print(f"Nama Mahasiswa : {nama_mhs}")
    print(f"Nilai Tugas    : {tugas}")
    print(f"Nilai Ujian    : {ujian}")
    print(f"Rata-rata      : {rata_rata}")
    print(f"Status Lulus   : {status_lulus}")
    print("-" * 35)

# 6. Uji program dengan minimal dua kombinasi nilai berbeda
print("--- UJI Kombinasi 1 ---")
rekap_nilai("Budi", 80, 85)

print("--- UJI Kombinasi 2 ---")
rekap_nilai("Siti", 70, 65)
```
